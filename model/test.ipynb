```python
import torch
def Binary_replace(indices,tgcots,chunk_size):
    '''
    tgcots is a list of tgcot,
    indices is a list of indice
    every tgcot in tgcots is a index list, divided into several chunks, and each chunk can be replaced by index in the indice or not, depending on the Binary distribution
    Binary distribution list is generated by torch.randint into a list, every position has 0.5 probability to be 1, 0.5 probability to be 0. if 1, replace the chunk with the index in the indice, if 0, keep the chunk as it is
    for example if chuck size = 2
        tgcot=[1,2,3,4,5,6,7,8]
        indice=[15,67,89,90]
        the Binary distribution is [0,1,0,1]
        then the rescot is [1,2,67,5,6,90]
    We want the rescots, a list of rescot
    '''
    rescots=[]
    binaries=[]
    for tgcot,indice in zip(tgcots,indices):
        binary=torch.randint(0,2,(len(tgcot)//chunk_size,)).tolist()
        binaries.append(binary)
        rescot=[]
        for i in range(len(binary)):
            if binary[i]==1:
                rescot+=indice[i:i+1]
            else:
                rescot+=tgcot[i*chunk_size:(i+1)*chunk_size]
        rescots.append(rescot)

    
    return rescots,binaries
    


tgcots=[[1,2,3,4,5,6,7,8],
        [7,3,4,8],
        [6,3,6,7,8,9,10,11,9,8]]
indices=[[15,67,89,90],
        [23,12],
        [67,35,89,90,12]]

rescots,binaries=Binary_replace(indices,tgcots,2)
print(rescots)
print(binaries)
```
